```python
import pandas as pd

df = pd.read_csv('/content/sample_data/test_reduzido.csv', index_col=False)
display(df.head(2))
```

```python
A = df['ParentalEducation'].unique()
B = df['SchoolType'].unique()
C = df['Locale'].unique()
print(A, B, C)
```

```python
locale_mapping = {
    'Suburban': 0,
    'City': 1,
    'Town': 2,
    'Rural': 3
}
df['Locale_numeric'] = df['Locale'].map(locale_mapping)


ParentalEducation_mapping = {
    'HS': 0,
    'Bachelors+': 1,
    'SomeCollege': 2,
    '<HS': 3,
}
df['ParentalEducation_numeric'] = df['ParentalEducation'].map(ParentalEducation_mapping)


SchoolType_mapping = {
    'Public': 0,
    'Private': 1,
}
df['SchoolType_numeric'] = df['SchoolType'].map(SchoolType_mapping)


display(df[['SchoolType', 'SchoolType_numeric', 'ParentalEducation', 'ParentalEducation_numeric', 'Locale', 'Locale_numeric']].head())

```

```python
df.drop('Locale', axis='columns', inplace=True)
df.drop('ParentalEducation', axis='columns', inplace=True)
df.drop('SchoolType', axis='columns', inplace=True)
```
